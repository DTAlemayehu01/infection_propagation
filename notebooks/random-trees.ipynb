# 2 Observer Histogram

```python
n = 1000
b = 6
h = Graph.random_tree_generator(n, b)
```

```python
h.birectional_simulate_gossip_rv('1', ['2','4'])
```

```python
times = []
for i in range(10**3):
    print(i)
    t = np.array(h.birectional_simulate_gossip_rv('1', ['2','4']))
    times.append(t)
times = np.array(times)
```

```python
xtimes = times[:, 0]
ytimes = times[:, 1]

hist, x_edges, y_edges = np.histogram2d(xtimes, ytimes, bins=20)

x_centers, y_centers = (x_edges[:-1] + x_edges[1:]) / 2, (y_edges[:-1] + y_edges[1:]) / 2

xpos, ypos = np.meshgrid(x_centers, y_centers, indexing="ij")

ax = plt.figure().add_subplot(111, projection="3d")

ax.bar3d(xpos.ravel() - (x_edges[1]-x_edges[0])/2, ypos.ravel() - (y_edges[1]-y_edges[0])/2,
    np.zeros(hist.size), x_edges[1]-x_edges[0], y_edges[1]-y_edges[0], hist.ravel())

ax.set(xlabel="dst[0] time", ylabel="dst[1] time", zlabel="Frequency", title="Infection time pairs vs frequency,n=1000,b=6")

plt.show()
```

# Parity Test

```python
n = 40
b = 4
p = 1.2*np.log(n)/n
def accuracy_trial(iters=10**3, edgedst=None):
    blind_times = []
    blind_paths = []
    informed_times = []
    informed_paths = []
    for i in range(iters):
        h = Graph.random_tree_generator(n,b,edge_dst=edgedst)
        # h = Graph.erdos_renyi(n,p,edge_dst=edgedst)
        blind_times.append(h.birectional_simulate_gossip_rv('1', '4'))
        p1 = h.construct_path('1','4')
        blind_paths.append(p1)
        h.sim_all()
        informed_times.append(h.algo_jump('1', '4', algorithm=""))
        p2 = h.construct_path('1','4')
        informed_paths.append(p2)
    return blind_times, blind_paths, informed_times, informed_paths
```

```python
res = accuracy_trial()
```
